import shutil
import os

parquet_path = "/home/jovyan/work/storage/historical_parquet"

if os.path.exists(parquet_path):
    try:
        # حذف المجلد بالكامل مع كل ملفات الباركيه المخزنة داخله
        shutil.rmtree(parquet_path)
        print(f"🧹 [نجاح] تم حذف وتطهير مجلد الباركيه بالكامل من المسار: {parquet_path}")
        print("🆕 المجلد الآن جاهز تماماً لاستقبال الباتش المعقم والنظيف بدون أي مخلفات قديمة!")
    except Exception as e:
        print(f"❌ فشل حذف المجلد: {e}")
else:
    print("ℹ️ المجلد غير موجود بالفعل، يمكنك البدء بالحقن مباشرة.")
